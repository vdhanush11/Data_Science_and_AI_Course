#### 4. What is exception handling? Give a real-life example of when it's useful.

* Exception handling is a programming technique used to detect and manage errors that occur during program execution.
* Instead of allowing a program to crash when an error occurs, exception handling enables the program to respond gracefully and continue running.

```python
# Example

try:
    user_input_1 = int(input("Enter your number:"))
    user_input_2 = int(input("Enter your second number:"))
    
    division = user_input_1/user_input_2
    print(division)
    
except ZeroDivisionError:
    print("You input is invalid, {} can't able to divide {}. so try to use more than zero".format(user_input_2,user_input_1))

except ValueError:                              
    print("You input is invalid, Try to use numbers as input.")                               

finally:
    print("Program Ended!!")
```

## 4. Given a string, count how many vowels it has.

```python
user_input  = input("Enter the string:").lower()
vowels      = "aeiou"
vowels_list = []

for i in user_input:
    if i in vowels:
        vowels_list.append(i)
print("Number of vowles:",vowels_list)
print("Count of vowles:",len(vowels_list))
        
```

## 9. Replace all missing values in a DataFrame with the column’s mean.
