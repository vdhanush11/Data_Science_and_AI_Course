# Number gussing game(Level 2-Using "for" & "while")

#### 1)"for" is used, when the number of iterations are known
#### 2)"while" is used, when the number of iterations are unknown or The instructions need to be excuted infinitelly
#### 3)"break" is used to stop the number of loops

```python
secrete_number=11

for i in range(0,7):
    user_input=int(input("enter your guessing:"))
    if secrete_number == user_input:
        print("congrates! your guess is correct")
        break
    else:
        print("sorry! your guess is wrong, Try again")
```

# Number gussing game(Level 3-Using multiple conditions)
#### 1."f" is called "print formating"
#### 2.Conditions:
#### 2.1 give number of attempts to guess
#### 2.2 give higher and lower
#### 2.3 give between 1-100, else give"Invalid Number"

```python
print("Welcome to the number guessing game")
print("Guess the number between 1-100, with 7 try's")

secret_number=11

for i in range(0,7):
    user_input=int(input("Enter your guess:"))
    if user_input<1 or user_input>100:
        print("Invalid number, your number should be between 1-100")
    elif user_input>secret_number:
        print("Your number is higher")
    elif user_input<secret_number:
        print("Your number is Lower")
    else:
        print("Congrats! You Guessed right")
        print(f"You guessed the {secret_number} with {i} attempts")
        break
    
```

# Number guessing game(Level 4-Using multiple conditions)
#### condition:
#### After completeing 7 tries, the program needs to print "Your 7 tries is over, Try Again"

```python
print("Welcome to the number guessing game")
print("Guess the number between 1-100, with 7 try's")

secret_number=11

for i in range(0,7):
    user_input=int(input("Enter your guess:"))
    if user_input<1 or user_input>100:
        print("Invalid number, your number should be between 1-100")
    elif user_input>secret_number:
        print("Your number is higher")
    elif user_input<secret_number:
        print("Your number is Lower")
    else:
        print("Congrats! You Guessed right")
        break
if secret_number == user_input:
    print(f"You guessed the {secret_number} with-in {i} attempts")
else:
    print("sorry!, All Your 7 tries is over, Try Again")
    
    
```

# Number guessing game(Level 6)
#### Condition
#### if user input is in string the  give waring ("can't able to understand!,used numbers as input between 1-100")

```python
import random

print("Welcome to the number guessing game")
print("Guess the number between 1-100, with 7 try's")

secret_number=random.randint(1,100)

while True:
    try:
        user_input=int(input("Enter your guess:"))
        if user_input<1 or user_input>100:
            print("Invalid number, your number should be between 1-100")
        elif user_input>secret_number:
            print("Your number is higher")
        elif user_input<secret_number:
            print("Your number is Lower")
        else:
            print("Congrats! You Guessed right")
            break
    except ValueError:
        print("can't able to understand!,used numbers as input between 1-100")
    
```
