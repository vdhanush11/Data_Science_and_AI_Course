## 3. Using user - define function
1. **Team_member** - should login, view task and submit task.
2. **Manager** - should login, view task and assign task.

```python
# Creating Function 

def login(user_input_1,user_input_2):
    print(f"{user_input_1['name']} has logged-in")
    print(f"{user_input_2['name']} has logged-in")

def view_task(user_input_1,user_input_2):
    print(f"{user_input_1['name']} has viewed task")
    print(f"{user_input_2['name']} has viewed task")
    
def submit_task(user_input_2):
    if user_input_2['role'] == "Team_member":
        print(f"{user_input_2['name']} has submitted task")
    else:
        print("Manager can't able to submit the task!")
    
def assign_task(user_input_1):
    if user_input_1['role'] == "Manager":
        print(f"{user_input_1['name']} has assigned task")
    else:
        print("Team members can't able to assign the task!")

#Data 
user_input_1 = {"name":"Dhanush", "role":"Manager"}
user_input_2 = {"name":"Ram","role":"Team_member"}


# Function to be called
login(user_input_1,user_input_2)
view_task(user_input_1,user_input_2)
submit_task(user_input_2)
assign_task(user_input_1)
```

## Same Conditions using OOPS

1. **Class** - Blueprint that contain attributes of a particular objects and it's functions.
2. **Object** - Instance of class.
3. **Inheritance** - It will inherite the parent class into the child class. We can easilly inherit the parent calss information into the child class.
4. **Encapsulasation** - It will used to hide the importsnt information to public eye.

```python
# class assigning

class Employee:
    def __init__(self,name):
        self.name = name

    def login(self):
        print(f"{self.name} is logged in.")
    def view_task(self):
        print(f"{self.name} is viewed task.")
        
class Team_member(Employee):
    def submit_task(self):
        print(f"{self.name} is submitted the task.")
        
class Manager(Employee):
    def assign_task(self):
        print(f"{self.name} is assigned task")
```

```python
# object assigning

dhanush = Team_member("dhanush")
chootabheem = Manager("chootabheem")
```

```python
# calling class 

dhanush.login()
dhanush.submit_task()
dhanush.view_task()
```

```python
chootabheem.assign_task()
chootabheem.login()
chootabheem.view_task()
```

```python
class SecretInformation:
    def __init__(self, name, atm_pin):
        self.name = name
        self.__atm_pin = atm_pin # encapsulation implemented.

    def get_pin(self):
        for i in range(3):
            pin = int(input("Enter ATM PIN: "))
            if pin == self.__atm_pin:
                print("PIN Verified")
                return
            else:
                print("Wrong PIN")
        print("Account Blocked")


user = SecretInformation("Dhanush", 1234)
user.get_pin()
```

## All four OOP pillars using one ATM

```python
# Abstraction

from abc import ABC, abstractmethod

class Bank(ABC):

    @abstractmethod
    def deposite(self):
        pass
    def withdrawal(self):
        pass
    def get_pin(self):
        pass

# Inheritance

class secure_data(Bank):
    def __init__(self,name,atm_pin):
        self.name = name
        self.__atm_pin = atm_pin   # Encapsulation(private)

# Polymorphism(Method overriding)

    def get_pin(self):
         for i in range(3):
             pin = int(input("Enter your pin:"))

             if pin == self.__atm_pin:
                 print("PIN VERFIED")
                 return
             else:
                 print("WRONG PIN")
    def deposite(self):
        amount = int(input("Enter amount to deposite:"))
        print(f"{amount} deposited successfully!.")
    def withdrawal(self):
        amount = int(input("Enter amount to withdrawal:"))
        print(f"{amount} withdrawed successfully!.")

# Object creation

user_data = secure_data("dhanush",2345)

user_data.get_pin()
user_data.deposite()
user_data.withdrawal()
    
```
