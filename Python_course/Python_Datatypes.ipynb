## Variable length argument
### Condition:
#### Suppose, You need to store the multiple arguments inside one parameter, Then use (*)

```python
#Suppose the expected salary is changing like 8,9,10

def client_from_chennai(name,place,*Expected_salary): # only one variable arguments should be there for one userdefine function.
    print("client Name:",name)
    print("client Expected salary:",Expected_salary)
    print("client place:",place)
```

```python
client_from_chennai("Dhanush","Chennai",7.5,9,10,12.5)
```

# Challenge
### Condition
#### Find the factors of a number

##### Ex: suppose, the user input is "20" then, the output should be "1,2,4,5,10,20"

```python
user_input=int(input("Enter the number to check factor:"))

for i in range(1,user_input+1):
    if user_input%i==0:
        print(i)
    
```

```python
# second method
user_input=int(input("Enter the number to check factor:"))

result=[]
for i in range(1,user_input+1):
    if user_input%i==0:
        result.append(i)
print("You factor for {} is {}".format(user_input,result))
    
```
