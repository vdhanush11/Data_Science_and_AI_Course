### 1. Groupby ()
* Discrite/catogorical - It is measurable and countable and does'nt have units.
* Continueous  - It will have units.
* When you need to group two column , then you need to use one continueous followd by, means ["region","sex"]["charges"] - Two discrite / one contineous.

```python
import pandas as pd
```

```python
Insurance_data = pd.read_csv("Insurance.csv")
Insurance_data.head()
```

#### Give the average charges from the each region.

```python
Insurance_data.groupby("region")["charges"].mean().round(2).sort_values()
```

#### Give the average charges from the each region and sex.

```python
Insurance_data.groupby(["region","sex"])["charges"].mean().round(2).sort_values(ascending = False)
```

### 2. pivot_table ()
* It will make the data's in more presentable in table.

```python
pd.pivot_table(data = Insurance_data,values="charges",index="region",columns=["sex","smoker"]).round(2)
```

### 3. Aggregate ()
* sum()
* min()
* max()
* mean()
#### 3.1 Suppose, I need to find the average,min,max,sum of charges with respect to region

```python
Insurance_data.groupby(["region","sex"])["charges"].agg(["mean","min","max","sum"]).round(2)
```

### 4. Cross_tab ()
* To compute the frequence of two or more catogorical features

#### 4.1 How many smoker and non - smoker are there based on region

```python
pd.crosstab(Insurance_data["region"], Insurance_data["smoker" ],margins = True)
```

### 5. concat ()

* It use to combine the different dataset into one dataset by appending one after another.
* we need to use this function when the different dataset will haves the same column or obervations name.  

```python
sales_17 = pd.read_csv("Sales-transaction-2017.csv")
sales_18 = pd.read_csv("Sales-transaction-2018.csv")
sales_19 = pd.read_csv("Sales-transaction-2019.csv")
```

```python
# I need to concat this three datasets.

sales_combined = pd.concat([sales_17,sales_18,sales_19])
sales_combined
```

```python
sales_combined.dtypes
```

#### Step 1 - Remove Duplicate

```python
sales_combined.drop_duplicates(inplace = True)
sales_combined
```

#### Step 2 - Fill missing values with mean()

```python
sales_combined.drop(columns = ["Gross","Disc","Voucher Amount"])
```

### 6. Apply ()

#### Create a new column with the name of "Eligible criteria" in that if age > 20 return eligible else not eligible

```python
def age_criteria(x):
    if x > 20:
        return "Eligible"
    else:
        return "Not Eligible"
```

```python
Insurance_data["Eligible Criteria"] = Insurance_data["age"].apply(age_criteria)
Insurance_data
```

### 7. lambda ()

Create a new column with the name of "Eligible criteria" in that if age > 20 return eligible else not eligible by using lambda in simple.

```python
Insurance_data["Eligible Criteria2"] = Insurance_data["age"].apply(lambda x: 1 if x>20 else 0)
Insurance_data
```

### 8. Delete () - del

```python
# It will delete one column at a time 

del Insurance_data["Eligible Criteria2"]
```
