# 1. List []

* List is 1D data structure, We can collect data at one dimension(row or column)
* Can allow Duplicate
* It is heterogenous means, It can allow to store different data types inside one list (int,float,boolean,str)
* The indexing is possible and it can denote using [], The indexing has two types (positive,negative),In positive indexing start from [0] for negative indexing start from [-1] in reverse.
* Slicing is possible.The slicing can be used by [:]
* It is Mutable mean, onces the values is assign to a variable and it can able to change or update in future.

```python
# Empty list

Empty_cart = []
```

```python
Empty_cart
```

```python
# Alow Duplicate

item_to_buy = ["apple","orange","graphs","watermelon","apple"]
print(item_to_buy)
```

```python
# Heterogenous

item_to_buy = ["apple","orange","graphs","watermelon",9,7.9,True]
print(item_to_buy)
```

```python
# Indexing (positive[0],negative[-1])

item_to_buy = ["apple","orange","graphs","watermelon",9,7.9,True]
print("positive indexing [2]:",item_to_buy[2])
print("negative indexing [-4]:",item_to_buy[-4])
```

```python
# Slicing

item_to_buy = ["apple","orange","graphs","watermelon",9,7.9,True]
print("positive Slicing [1:3]:",item_to_buy[1:4])
print("negative Slicing [-4:-1]:",item_to_buy[-4:-1])
```

```python
# Mutable (Updatable)

item_to_buy = ["apple","orange","graphs","watermelon",9,7.9,True]
item_to_buy[1] = "Pineapple"
print(item_to_buy)
```

# 2. Sets {}

* You can present data in 1D
* Sets will not allow Duplicate values
* Sets is heterogeous
* sets can't able to perform indexing & slicing
* It is mutable using build-it fuction of "sets.add()"
* It will give first priorty to numbers and then to alphabetic, It will arrange {float -> int -> alpahbetic}

```python
# Empty sets

Empty_cart = {}
```

```python
Empty_cart
```

```python
# Can't Alow Duplicate

item_to_buy1 = {"apple","orange","graphs","watermelon","apple"}
item_to_buy1
```

```python
# It is heterogenous

item_to_buy1 = {"apple","orange","graphs","watermelon","apple",6.75,9}
item_to_buy1
```

```python
# Mutable using Build-in fuction

item_to_buy1 = {"apple","orange","graphs","watermelon","apple",6.75,9}
item_to_buy1.add("banana")
item_to_buy1
```
