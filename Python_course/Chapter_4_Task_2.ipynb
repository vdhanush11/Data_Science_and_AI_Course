## Task 1 - Data Cleaning Challenge on Sales Data.

```python
import pandas as pd
```

```python
# combining three datasets using concat()

sales_17 = pd.read_csv("Sales-transaction-2017.csv")
sales_18 = pd.read_csv("Sales-transaction-2018.csv")
sales_19 = pd.read_csv("Sales-transaction-2019.csv")
```

```python
sales_combined = pd.concat([sales_17,sales_18,sales_19])
sales_combined
```

```python
sales_combined.drop_duplicates(inplace = True)
sales_combined
```

#### Step 2 - Checking missing values and replacing NaN with 0

```python
sales_combined.isna().sum()
```

```python
sales_combined["Voucher Amount"] = sales_combined["Voucher Amount"].fillna(0)
sales_combined["Disc"]           = sales_combined["Disc"].fillna(0)
```

```python
sales_combined
```

#### Step 3 - Remove all rows where any of these columns contain missing values (NaN).

```python
sales_combined.dropna(subset=["Date", "Voucher", "Party", "Product", "Qty", "Rate", "Gross"],inplace = True)
```

## Task 2 - Merge Function 1 Example
1. inner merge
2. outer merge
3. left merge
4. right merge
5. cross merge

```python
sales_17 = pd.read_csv("Sales-transaction-2017.csv")
insurance_details = pd.read_csv("Insurance.csv")
```

```python
sales_17.head()
```

```python
sales_18.head()
```

```python
sales17_summary = sales_17.groupby("Product")["Qty"].sum().reset_index()

sales18_summary = sales_18.groupby("Product")["Qty"].sum().reset_index()

# Inner Merge
inner_merge = pd.merge(sales17_summary, sales18_summary,on="Product", how="inner",suffixes=("_2017", "_2018"))

# Left Merge
left_merge = pd.merge(sales17_summary, sales18_summary,on="Product", how="left",suffixes=("_2017", "_2018"))

# Right Merge
right_merge = pd.merge(sales17_summary, sales18_summary,on="Product", how="right",suffixes=("_2017", "_2018"))

# Outer Merge
outer_merge = pd.merge(sales17_summary, sales18_summary,on="Product", how="outer",suffixes=("_2017", "_2018"))
```

```python
inner_merge.head()
```

```python
outer_merge.head()
```

```python
right_merge.head()
```

```python
left_merge.head()
```

```python
# Cross merge

cross_merge = pd.merge(sales17_summary.head(10),sales18_summary.head(10),how="cross",suffixes=("_2017", "_2018"))
cross_merge.head()
```

```python
# Merge with indicator

indicator_merge = pd.merge(sales17_summary,sales18_summary,on="Product",how="outer",indicator=True,suffixes=("_2017", "_2018"))
indicator_merge.head()
```

## Task 3 - List Comprehension 2 examples

#### Example 1 - Find Passed Students

```python
student_marks = [30,40,60,90,26,39,70,100]

passed_studend = [i for i in student_marks if i >= 40]

print("Passed students:",passed_studend)
```

#### Example 2 - Find employees assigned to active projects.

```python
import pandas as pd
```

```python
employees                  = ["Dhanush","Ram","Kumar","Sundar","Virat"]
projects                   = ["AI Project", "Web App", "Testing","Model bulding","Data cleaning"]
Inactive_employees         = ["Kumar","Ram"]

project_assigned_employees = [[emp,"No project" if emp in Inactive_employees else proj,"InActive" if emp in Inactive_employees else "Active"] 
                              for [emp,proj] in zip(employees,projects)]

pd.DataFrame(data = project_assigned_employees)
```

## Task 4 - Lambda Function 2 examples with Map, Filter, Reduce


#### 1. MAP – Increase charges by 10%

```python
insurance_data_charges = pd.read_csv("Insurance.csv")
insurance_data_charges
```

```python
insurance_data_charges["charges"] = list(map(lambda charge: charge * 1.10,insurance_data_charges["charges"]))
insurance_data_charges
```

#### 2. FILTER - Find customers whose insurance charges are greater than ₹15,000.

```python
insurance_data_charges["charges"] = list(map(lambda charge: charge > 15000,insurance_data_charges["charges"]))
insurance_data_charges
```

#### 3. REDUCE – Find the Highest goal scored by a team.

```python
from functools import reduce
```

```python
euro_football = pd.read_csv("Euro_2012.csv")
euro_football.head()
```

```python
highest_goal_team = reduce(lambda x, y: x if x["Goals"] > y["Goals"] else y,euro_football.to_dict("records"))

pd.DataFrame(
    {
        "Team": [highest_goal_team["Team"]],
        "Goals": [highest_goal_team["Goals"]]
    }
)
```
