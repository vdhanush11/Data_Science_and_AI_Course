# Task 3
#### Rock, Paper, Scissors Game

#### Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.


```python
print("WELCOME TO THE ROCK, PAPER, SCISSOR GAME!!")

import random

while True:
    try:
        user_input = input("Enter the choice (rock/paper/scissor):").lower()

        if user_input.isdigit():
            raise ValueError

        if user_input not in ["rock","paper","scissor"]:
            print("\nInvalid text!, Please select only one among this (rock/paper/scissor).")
            continue

        computer_choice = ["rock","paper","scissor"]
        computer_choice = random.choice(computer_choice)
        
        if computer_choice == user_input:
            print("IT'S A TIE!")    
        elif computer_choice == "rock" and user_input == "paper" or computer_choice == "scissor" and user_input == "rock" or computer_choice == "paper" and user_input == "scissor":
            print("USER WON!")
        else:
            print("COMPUTER WON!")
        
        play_again = input("\nIf you want to play again? (yes/no)").lower()
        if play_again != "yes":
            print("Thanks for playing")
            break
    except ValueError:
        print("\nInvalid input!, Try to enter in alphabetical letters")

```

# Task 4 Word Guessing Game (Mini Hangman)
#### Instructions:
1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2.	Randomly select one word for the game.
3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4.	Keep track of wrong guesses (max 6 attempts).
5.	End the game when the player either guesses the full word or uses all chances.


```python
import random

words_list = ["python", "machine", "learning", "bridge"]
select_word = random.choice(words_list)

guessed_letters = []

for i in range(0,6):

    display = ""

    for letter in select_word:
        if letter in guessed_letters:
            display += letter
        else:
            display += "_"

    print("\nWord:", display)

    user_input = input("Enter your guess: ").lower()

    if user_input in select_word:
        guessed_letters.append(user_input)
        print("Correct Guess!")

        display = ""

        for letter in select_word:
            if letter in guessed_letters:
                display += letter
            else:
                display += "_"

        if display == select_word:
            print("YOU WON!")
            break

    else:
        print("Wrong Guess!")

else:
    print("\nGAME OVER!")
    print("The word was:", select_word)
```
