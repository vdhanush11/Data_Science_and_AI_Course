## 1) List - []

```python
# remove() items from the list

item_to_buy_1 = ["apple","orange","graphs","watermelon","Jack Fruit"]
item_to_buy_1.remove("apple")

item_to_buy_1
```

```python
# append() items to the list

item_to_buy_2 = ["apple","orange","graphs","watermelon","Jack Fruit"]
item_to_buy_2.append("banana")

item_to_buy_2
```

```python
# count() number of items in the list

item_to_buy_3 = ["apple","orange","graphs","watermelon","Jack Fruit","apple"]
print(item_to_buy_3.count("apple"))
```

```python
# sort() the list of items

item_to_buy_4 = ["apple","orange","graphs","watermelon","jackFruit","apple"]
item_to_buy_4.sort()
print(item_to_buy_4)
```

```python
# reverse() the list of items

item_to_buy_5 = ["apple","orange","graphs","watermelon","jackFruit"]
item_to_buy_5.reverse()
print(item_to_buy_5)
```

```python
# Using pop() function to remove the perticular values using it's index value.

item_to_buy_6 = ["apple","orange","graphs","watermelon","jackFruit"]
item_to_buy_6.pop(2)
print(item_to_buy_6)
```

```python
# Using insert() function to insert the value using the before index value from the list

item_to_buy_7 = ["apple","orange","graphs","watermelon","jackFruit"]
item_to_buy_7.insert(2,"Banana")
print(item_to_buy_7)
```

```python
# Using index() function to return the index value of the perticular item from the list.

item_to_buy_8 = ["apple","orange","graphs","watermelon","jackFruit"]
print(item_to_buy_8.index("watermelon"))
```

```python
# Using extend() function to extend the list of items

item_to_buy_9 = ["apple","orange","graphs","watermelon","jackFruit"]
item_to_buy_9.extend(["pine apple","dragon fruit"])
print(item_to_buy_9)
```

```python
# Using copy() function for making the copy list of original list

item_to_buy_10 = ["apple","orange","graphs","watermelon","jackFruit"]
copy_of_list = item_to_buy_10.copy()
print(copy_of_list)
```

```python
# Using clear() function to turminate the list of items

item_to_buy_11 = ["apple","orange","graphs","watermelon","jackFruit"]
item_to_buy_11.clear()
print(item_to_buy_11)
```

## Sets - {}

```python
# Using add() to add values to the sets

client_name_1 = {"dhanush","ram","bheem","jaggu"}
client_name_1.add("motu")
print(client_name_1)
```

```python
# Using update() to update values to the sets

client_name_2 = {"dhanush","ram","bheem","jaggu"}
client_name_2.update({"dolu","bolu"})
print(client_name_2)
```

```python
# Using remove()

client_name_3 = {"dhanush","ram","bheem","jaggu"}
client_name_3.remove("bheem")
print(client_name_3)
```

```python
# Using pop()

client_name_4 = {"dhanush","ram","bheem","jaggu"}
client_name_4.pop()
print(client_name_4)
```

```python
# Using copy()

client_name_5   = {"dhanush","ram","bheem","jaggu"}
new_client_list = client_name_5.copy()
print(new_client_list)
```

```python
# Using clear()

client_name_6 = {"dhanush","ram","bheem","jaggu"}
client_name_6.clear()
print(client_name_6)
```

```python
# Using union()

client_name_7 = {"dhanush","ram","bheem","jaggu"}
new_client    = {"dolu","bolu"}
updated_list  = client_name_7.union(new_client)
print(updated_list)
```

```python
# Using difference()

client_name_8 = {"dhanush","ram","bheem","jaggu"}
new_client    = {"ram","jaggu"}
updated_list  = client_name_8.difference(new_client)
print(updated_list)
```

```python
# Using difference_update()

client_name_9 = {"dhanush","ram","bheem","jaggu"}
new_client    = {"ram","jaggu"}
client_name_9.difference_update(new_client)
print(client_name_9)
```

```python
# Using intersection_update()

client_name_10 = {"dhanush","ram","bheem","jaggu"}
new_client    = {"ram","jaggu"}
client_name_10.intersection_update(new_client)
print(client_name_10)
```

```python
# Using intersection()

client_name_11 = {"dhanush","ram","bheem","jaggu"}
new_client    = {"ram","jaggu"}
updated_list  = client_name_11.intersection(new_client)
print(updated_list)
```

```python
# Using discard()

client_name_12 = {"dhanush","ram","bheem","jaggu"}
client_name_12.discard("ram")
print(client_name_12)
```

```python
# Using isdisjoint() - checking any common values between two variables

client_name_12 = {"dhanush","ram","bheem","jaggu"}
new_client     = {"vijay","rajini"}
print(client_name_12.isdisjoint(new_client))
```

```python
# Using issubset() - checks whether all values from set1 are equals the set2

client_name_12 = {"dhanush","ram","bheem","jaggu"}
new_client     = {"vijay","rajini"}
print(client_name_12.issubset(new_client))
```

```python
# Using issuperset()

client_name_12 = {"dhanush","ram","bheem","jaggu"}
new_client     = {"dhanush","bheem"}
print(client_name_12.issuperset(new_client))
```

```python
# Using symmetric_difference()

client_name_12 = {"dhanush","ram","bheem","jaggu"}
new_client     = {"dhanush","bheem"}
print(client_name_12.symmetric_difference(new_client))
```

```python
# Using symmetric_difference()

client_name_13 = {"dhanush","ram","bheem","jaggu"}
new_client     = {"dhanush","bheem"}
client_name_13.symmetric_difference_update(new_client)
print(client_name_13)
```

## Tuple ()

```python
# Using count() - To find number of duplicate values in tuple.

store_product = ("iphone18","airpods","laptop","mouse","laptop")
print(store_product.count("laptop"))
```

```python
# Using index() - To find exact location using index value
 
store_product = ("iphone18","airpods","laptop","mouse","laptop")
print(store_product.index("mouse"))
```

## Dictionary {}

```python
# Store customer details

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details
```

```python
# Using keys() 

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.keys()
```

```python
# Using values() 

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.values()
```

```python
# Using update() 

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.update({"Amount_paid":81000})
print(customer_details)
```

```python
# Using pop() 

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.pop("Name")
print(customer_details)
```

```python
# Using update() 

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.update({"Amount_paid":81000})
print(customer_details)
```

```python
# Using popitem() - To remove last key value pairs

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.popitem()
print(customer_details)
```

```python
# Using setdefault()

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.setdefault("city","vellore")
print(customer_details)
```

```python
# Using items()

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.items()
print(customer_details)
```

```python
# Using get()

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
print(customer_details.get("Amount_paid"))
```

```python
# Using fromkeys() to set default values

fields = ["Name", "Age", "City", "Mobile"]
customer_form = dict.fromkeys(fields, "Pending")
print(customer_form)
```

```python
# Using copy()

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.copy()
print(customer_details)
```

```python
# Using clear()

customer_details = {"Name":"Dhanush","Age":22,"product_purchased":"Laptop","Amount_paid":72000}
customer_details.clear()
print(customer_details)
```
