## What is Python Library ?

#### It is a pre-written code which contains (modules,functions and class) that you can use to perform a tasks without writting everything from scratch.

#### Saves time      - No need to write big code for common tasks

### 1. Pandas - Used for Data Analysis


```python
# Going to read csv file (comma seperated values).

import pandas as pd #(pd) is alies like nick name.
```

```python
employee_details = pd.read_csv(filepath_or_buffer = "sample-simple.csv")
```

```python
employee_details
```

## 2. Numpy - Used in numerical operation

```python
import numpy as np
```

```python
# repeat() function is used to duplicate the information inside the dataset.

np.repeat(a=employee_details,repeats=2,axis=0) # axis will operate the row or column.
```

```python
# Now i need it proper structure formate so, iam using pandas.

# axis = 0: it will see the row , axis = 1: it will see the column

pd.DataFrame(data = np.repeat(a=employee_details,repeats=2,axis=0) )
```

## 3. Matplotlib, seaborn, plotly - Used for data visualization

#### Two types of datatypes

### Matplotlib

```python
from matplotlib import pyplot as plt
```

```python
plt.hist(employee_details.age) # we can't able to visualized the whole datasets, we can only access perticular column of dataset
```

### Seaborn

```python
import seaborn as sns
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = employee_details[['id', 'age', 'salary']]

corr = numeric_cols.corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```
