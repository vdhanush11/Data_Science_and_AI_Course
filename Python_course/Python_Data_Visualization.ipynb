### 3.2 Heatmap

#### Pearson corelation , co-efficiant
1. It will represent the data from **-1** to **+1**.
2. When x and y axis are increases, Then it is **positivelly corelated**.
3. If x_axis increases and y_axis decreses, Then it is **negativelly correlated**.

### To represent the corelation matrix in attractive way, Then we can use heatmap using seaborn.

```python
# TO REPRESENT THE CORRELATION MATRIX , THEN DATASET SHOULD CONTAIN ONLY NUMERICAL VALUES. SO IAM GOING TO DEL THE STRING COLUMNS.

del mtcars_data["model"]
```

```python
mtcars_data.head()
```

```python
mt_correlation_matrix = mtcars_data.corr().round(2)
mt_correlation_matrix
```

```python
# To represent this more effictive way, I'm using heat map.

sns.heatmap(data = mt_correlation_matrix, annot = True)
plt.show()

# INSIGHTS:
# Here milage(mpg) is very negatively correlated with respect to cylinder(cly).
```

### Notes
1. **0 to 0.20** = VERY WEAKLY CORRELATED.                                                  
3. **0.21 to 0.40** = WEAKLY CORRELATED.
4. **0.41 to 0.60** = NORMALY CORRELATED.
5. **0.61 to 0.80** = STRONGLY CORRELATED.
6. **0.81 to 0.1** = VERY STRONGLY CORRELATED
