### 1. Create a Python program to simulate a simple ATM system using Object-Oriented Programming (OOP).

**I'm goig to use four main pillers of OOPS**
1. Abstraction
2. Inheritance
3. Encapsulation
4. Polymorphism

```python
# ABSTRACTION

from abc import ABC, abstractmethod

class Bank(ABC):
    def check_balance():
        pass
    def deposite_amount():
        pass
    def withdrawl_amount():
        pass
    def get_atm_pin():
        pass

# INHERITANCE

class secure_data(Bank):
    def __init__(self,name,atm_pin):
        self.name    = name
        self.__atm_pin = atm_pin  # ENCAPSULATION (Private)
        self.balance = 10000      # Initial balance

# POLYMORPHISM (Overriding)

    def get_atm_pin(self):
        for i in range(0,3):
            pin = int(input("Enter you ATM pin:"))
            if pin == self.__atm_pin:
                print("PIN VERIFIED")
                return
            else:
                print("WRONG PIN")

    def deposite_amount(self):
        amount = int(input("Enter amount to deposite:"))
        self.balance +=amount
        print(f"{amount} IS DEPOSITED SUCCESSFULLY!")

    def withdrawl_amount(self):
        amount = int(input("Enter amount to windrawal:"))
        if amount <= self.balance:
            self.balance -= amount
            print(f"{amount} IS WITHDRAWLED SUCCESSFULLY!")
        else:
            print("INSUFFICIANT BALANCE")

    def check_balance(self):
        print("---AVILABLE BALANCE---")
        print(f"{self.balance} IS YOU AVAILABLE BALANCE")
        
```

```python
# OBJECT CREATION

user_data = secure_data("name",2345)

# PERFORMING TASKS

user_data.get_atm_pin()
user_data.deposite_amount()
user_data.withdrawl_amount()
user_data.check_balance()
```
